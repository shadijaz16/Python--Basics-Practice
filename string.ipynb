# STRING FUNCTION

### STRING:

```python
*A string is a sequence of characters used to store text data in python
*strings are written inside quotes
eg:
    'hello'
    "hello"
  '''hello'''
  """hello"""
*strings are immutable (cannot be changes after creation)
*each character has an index position.
*can contain letters,numbers,symbols and spaces
eg:
  'a'
  '2'
  '*'
  ''
```

```python
name='diya'
age='25'
college='najath college of arts and science karuvarakundu'
email_id='diyashameer790@gmail.com'
city= 'malappuram'
degree='bcom coop'
print(name)
print(age)
print(college)
print(email_id)
print(city)
print(degree)
```
