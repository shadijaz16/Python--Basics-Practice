#### 4.Arbitrary Arguments.(*args)
Used when the number of arguments is unknown.
 We have to add one * before the name of parameter.

```python
def s(*a):
    print(sum(a))
s(2,9,8,7,6,3,4,1)
```

```python
def s(*a):
    return sum(a)
sum=s(2,9,8,1,4,6,3)
print("total",sum)
```

## Recursion

```python
What is Recursion?
Recursion is a programming technique where a function calls itself repeatedly until 
a base condition is met. It is useful for solving problems that can be broken down 
into smaller subproblems of the same type.

Key Components of Recursion
Base Case – The stopping condition to prevent infinite recursion.
Recursive Case – The function calls itself with a smaller or simpler input.
```

### Base Case: if n==1,return 1. 
Recursive Case :n*factorial(n-1) reduces n by 1 until the base case is met.

```python
#explanation
Recursive function: factorial n calls itself with n-1 until it reaches the base case.
Base case: When n-- 1, the function returns 1.
Recursive case: factorial(n) = n * 
```

## Factorial of 5

```python
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n - 1)
print(factorial(5))
```

```python
def fibonacci(n):
    if n==0 or n==1: #base case
        return n
    return fibonacci(n-1) + fibonacci(n-2) #recursive call
print(fibonacci(6))
        
```
