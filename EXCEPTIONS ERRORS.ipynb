### EXCEPTION HANDLING

```python
Python provides a handling system for handling errors through exception handling
```

```python
SYNTAX

try:
    # code that might raise an exception
except:
    # Hndle the exception
else:
    # code that runs if no exeption occurs (optional)
    finally:
    # code that runs no matter what (optional)
```

```python
 try Block:

Code that might raise an error is placed inside a try block.
If an error occurs, the flow of control jumps to the corresponding except block.

except Block:

This block catches and handles the exception raised in the try block.
You can catch specific exceptions or use a general exception handler.

else Block:

The else block runs only if no exception was raised in the try block.
It is useful when you need to run code only when everything goes smoothly.

finally Block:

The finally block is executed no matter what, whether an exception was raised or not.
This block will always executed.
```

```python
Question:
Write a program that asks the user to enter a number,
👉 checks whether the number is positive,
👉 and handles any exception using a general except block,
👉 finally print "Execution completed".
```

```python
try:
    num=int(input("enter a number : "))
    if num>0:
        print("its a positive number")
    elif num==0:
        print("its zero")
        
    else:
        print("its negative number")
        
except ValueError:
    print("invalid number")
finally:
    print("Execution completed")
```
