## FOR LOOP

```python
A for loop is used to repeat a block of code for each item in a sequence.
    sequence--->list,tuple,string,range
```

```python
syntax:
for variable in sequence:
    statements
```

```python
#1.sequence=list
a=[1,2,3,4]
for i in a:
    print(i)
    
```

```python
#sequence list:
a=["apple","kiwi","mango"]
for i in a:
    print(i)
```

```python
#3.sequence string:
a="diya"
for i in a:
    print(i)
```

```python
#4.sequence in tuple:
a=("kiwi","apple","mango")
for i in a:
    print(i)
```

```python
#5.range--range(start,stop,step)
for i in range(1,11):
    print(i)

```

```python
# odd numbers
for i in range(1,11,2):
    print(i)
```

```python
#even numbers
for i in range(2,11,2):
    print(i)
    
```

## LOOPING CONTROL STATEMENTS

### 1.BREAK

```python
if break used it will stop at the given string. 
```

```python
fruits=["mango","kiwi","apple"]
for i in fruits:
    print(i)
    if i =="kiwi":
        break
```

### 2.continue

```python
it will skip the given string and continues.
```

```python
fruits=["mango","kiwi","apple"]
for i in fruits:
    if i =="kiwi":
        continue
    print(i)
```

### 3.pass

```python
when pass uses it will not get the output and its doesnt display the error.
```

```python
fruits=["mango","kiwi","apple"]
for i in fruits:
    pass
```

### QN:check whether a number is amstrong number

```python
num=input('num')
sum=0
for i in num:
    cube=int(i)**3
    sum=sum+cube
if sum==int(num):
    print("its an amstong number")
else:
    print("its not")
```
