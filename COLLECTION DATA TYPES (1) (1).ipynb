# 3.SET

```python
a set is a collection data tyupe used to store
multiple values like list and tuple,but without duplicates.
```

```python
features:
*unordered
*no indexing
*mutable
*does not allow duplicates
*stores unique elements only
```

```python
a={"a","b","c"}
a
```

```python
#remove repeated elements from a list by coverting into set.
a=[1,2,2,3,3]
b=set(a)
c=list(b)
c
```

```python
#integer set
a={1,2,5,8,9,4}
print("integer set:",a)
#string set
b={"a","b","c"}
print("string set:",b)
#float set
c={2.5,6.2,8.7}
print("float set:",c)
#multiple set
d={1,2,"b","c",6.2,8.7}
print("mixed set:",d)
```

## Set operations:

```python
set operations are used to compare,compine or filter elements from 2 or more sets.
```

```python
1.union
2.intersection
3.difference
4.symmetric_difference
5.intersection_update
6.difference_update
7.symmetric_difference_update
```

### 1.UNION:('|')

```python
combine all elements from both sets(no duplicates)
```

```python
a={1,2,3}
b={4,5,6}
print(a.union(b)) #method
a|b               #operator
```

### 2.intersection: (&)

```python
common elements from both sets.
```

```python
a={1,2,3,5}
b={4,5,6,1}
print(a.intersection(b))
a&b
```

### 3.Difference: ('-')

```python
elements in first set but not in the second one
```

```python
a={1,2,3,5}
b={4,5,6,1}
print(a.difference(b))
a-b.
```

### 4.symmetric_difference ('^')

```python
elements in either set but not in both.
```

```python
a={1,2,3,5}
b={4,5,6,1}
print(a.symmetric_difference(b))
a^b
```

### 5.intersection_update ('&=')

```python
a={1,2,3,5}
b={4,5,6,1}
a.intersection_update(b)
a&=b
a
```

### 6.Difference update: ('-=')

```python
a={1,2,3,5}
b={4,5,6,1}
a.difference_update(b)
print(a)
a-=b
a
```

### 7.symmetric_difference_update: ('^=')

```python
a={1,2,3,5}
b={4,5,6,1}
a.symmetric_difference_update(b)
print(a)
a^=b
a
```
