### EXTRA Qstn ANAGRAM,ISOGRAM,FIBONACCI

```python
#Anagram---->there must be contain same element in a word.
a=input("enter a word")
b=input("enter second word")
sort1=sorted(a)
sort2=sorted(b)
if sort1==sort2:
    print("its an anagram")
else:
    print("its not")
```

```python
#Isogram---->it doesnt take duplicates in an element.
a = input("enter a word:")
b = set(a)
if len(a) == len(b):
    print("its an isogram")
else:
    print("its not")
```
