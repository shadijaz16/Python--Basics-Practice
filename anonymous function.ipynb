## 3.ANONYMOUS FUNCTIONS:

```python
*an anonymous function is a function without a name, also known as a lambda function.
    
*syntax:    
"lambda arguments: expression"

lambda -----defines the function.
arguments ----are the inputs.
expression -----is evaluated and returned.
    
*function doesnt have any name but this function can be assign into a perticular variable.
*defined by 'lambda' keyword
*It can have any number of arguments, but only one expression.
*It is mainly used for short, simple functions.
*do not defined by def keyword
*return expression but not value
*its a one line functions
*cannot access global variables
```

```python
SQUARE OF A NUMBER
```

```python
# by using def
def square(a):
    return a**2
print(square(5))
```

```python
# by usind lambda
#'lambda arguments:expression"
square = lambda a:(a**2)
print(square(5))
```

## 1. map() with lambda
The map() function applies a given function to each item in an iterable
(like a list or tuple) and returns a new iterable (map object).

```python
syntax:

map(function, iterable)
```

## EXAMPLE : SQUARRING A LIST

```python
a=[1,2,3,5,7]
sqr=list(map(lambda a:a**2,a))
sqr
```

```python
a=(1,4,6,8)
sqr=tuple(map(lambda a:a**2,a))
sqr
```

## 2.filter() with lambda

#### filter() – it is used to filter elements based on a condition
(i.e., returns only the elements that satisfy the condition).

```python
syntax:
filter(function,iterable)
```

## eg; FILTERING EVEN NUMBERS

```python
a=[1,2,3,4,5]
even=list(filter(lambda a:a % 2==0,a))
even
```

## EG; FILTERING ODD NUMBERS

```python
a=[1,2,3,4,5]
odd= list(filter(lambda a:a % 2==1,a))
odd
```
