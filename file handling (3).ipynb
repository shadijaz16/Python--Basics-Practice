# FILE HANDLING IN PYTHON

```python
In Python, creating and handling files is simple. 
Here's how you can create a new file and handle it
(i.e., read, write, append, or manipulate the contents). 
I'll explain each step thoroughly.
```

```python
Summary of File Handling in Python:

1.Create a File: Use open("filename", "w") to create a file or overwrite it.
    
2.Write to a File: Use the write() method to add content. Use 'a' mode to append
without overwriting.
    
3.Read from a File: Use read(), readline(), or readlines() to read content from the file.

4.Close the File: Always close the file using file.close() after reading or writing. 
Or, use with for automatic closure.

5.Use with: It ensures that files are properly closed after use, even in the case of errors.
```

```python
1. Creating a File
To create a file, you use Python's built-in open() function with the mode 'w' (write) or
'x' (exclusive creation).

'w' mode: If the file doesn't exist, it will be created. 
If it already exists, the content will be overwritten.
'x' mode: If the file already exists, it will raise an error. 
    It is used to ensure a file is created only if it doesn't already exist.
```

```python
2. Writing to a File
Once the file is created, you can write content to it using the write() method.

write(): This method writes a string to the file. 
It does not automatically add a newline at the end, so you need to use \n for new lines.
```

```python
# Open the file in write mode ('w')
file = open("techolastech.txt", "w")

# Writing content to the file
file.write("Hello, World!\n")
file.write("This is a new file created in Python.\n")
file.write("python is an interpreted language.\n")

# Close the file
file.close()
```

```python
3. Appending to a File
If you want to add content to an existing file without overwriting it, 
you can open the file in append mode ('a').
```

```python
f=open("techo.txt","w")
f.write("dilna.\n")
f.write("hiba.\n")
f.close()
```

```python
file = open("karnik.txt", "a")  # Open file in append mode
file.write("ansil.\n")  # Append text
file.close()  # Close the file
```

```python
Example: Appending content to a file
```

```python
# Open the file in append mode ('a')
file = open("subadra.txt", "a")

# Appending new content
file.write("This is appended text.\n")
file.write("Another line of text.\n")

# Close the file
file.close()
```

```python
4. Reading from a File
To read content from a file, you can open it in read mode ('r') using read(), readline(),
or readlines().

read(): Reads the entire content of the file.
readline(): Reads a single line at a time.
readlines(): Reads all lines of the file into a list where each element is a line.
```

```python
file = open("subadra.txt", "r")  # Open file in read mode
c = file.readlines()  # Read all content  # Close the file
print(c)  # Print the content
file.close()
```

```python
Example: Reading the entire file content
```

```python
# Open the file in read mode ('r')
file = open("jalal.txt", "r")

# Read and print all contents
content = file.readlines()
print(content)

# Close the file
file.close()
```

```python
Example: Reading file line by line
```

```python
# Open the file in read mode ('r')
file = open("example.txt", "r")

# Reading the file line by line
for line in file:
    print(line.strip())  # .strip() removes leading/trailing whitespace

# Close the file
file.close()
```

```python
with open("example.txt", "r") as file:
    lines=file.readlines()[:5]
    print("".join(lines))
    file.close()
```

```python
6. Using with for File Handling (Recommended)
Using with ensures that the file is automatically closed after the block of code is executed, even if there are errors.
```

```python
why use with? 
*It automatically handles the closing of the file, 
*so you don't need to explicitly call file.close(), making the code cleaner and more error-proof.
```

```python
Example with Writing:
```

```python
# Using 'with' to open and write to the file
with open("example.txt", "a") as file:
    file.write("This is a file written using 'with'.\n")
    file.write("It's easier and safer to use.\n")
   
```

```python
Example with Reading:
```

```python
# Using 'with' to open and read the file
with open("exp.txt", "r") as file:
    content = file.read()
    print(content)
```
