## creating 1 DIMENSIONAL ARRAY

```python
#list
a=[2,5,5]
print(type(a))
print(a)

#array using tuple
b=np.array((5,6,5))
print(type(b))
print(b)

#array using list
c=np.array([51,36,50])
print(type(c))
print(c)

#array using set
d=np.array({51,36,50})
print(type(d))
print(d)

```

## ADDING ELEMENTS TO AN ARRAY

```python
#1.append()----->adding an element at the end
```

```python
a=np.array([10,20,30])
print(a)
b=np.append(a,40)
b
```

```python
#2.to add the multiple elements at the end
```

```python
a=np.array([10,20,30])
print(a)
b=np.append(a,[40,10])
b
```

```python
#3.insert()--->add elements at specific possition.
#syntax:np.insert(variable,possition,value)
```

```python
a=np.array([10,20,30])
print(a)
b=np.insert(a,1,40)
b
```

## removing elements:

```python
#1.delete()--->remove specific value from a array
```

```python
b=np.array([10,20,30,40])
print(b)
c=np.delete(b,1)
c
```

```python
#1.to delete multiple elements using DELETE()--->remove specific value from a array
```

```python
b=np.array([10,20,30,40])
print(b)
c=np.delete(b,[1,3])
c
```

## 3 DIMENSIONAL ARRAY

```python
ITS A COLLECTION OF 2 DIMENSIOANAL ARRAYS.(MULTIPLE MATRIXES)
```

```python
A 3D ARRAY IS A COLLECTION OF 2 D ARRAYS.
```

```python
a=np.array([[[3,6,7],[5,8,9]],[[2,6,8],[6,8,9]],[[2,3,4],[3,2,4]]])
print(a)
```

```python
b=np.array([[[5,2,4],[8,4,1],[9,2,1]],[[7,1,4],[2,1,4],[3,2,1]]])
print(b)
b.shape
```
