```python
5.IDENTITY OPERATOR
```

```python
These are used to check whether 2 variables refer
to the same object in memory,not whether they have same value.
```

```python
a=[1,2,3]
b=a
c=[1,2,3]
print(a is not c)
print(a is c)
```

```python
python reuses memory for small integers(-5 to 256)-integer catching so both a and b
point to the same memory object.
```

```python
#eg
a=2
b=a
c=2
a is c
```

```python
a=257
b=a
c=257
a is c
```
